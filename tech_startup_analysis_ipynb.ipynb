```python
# CREATING A COPY OF THE DATASET
df_clean = df.copy()
```

```python
# CHECKING FOR DUPLICATES 
print("Duplicates:", df_clean.duplicated().sum())
```

```python
# REMOVING DUPLICATES 
df_clean.drop_duplicates()
```

```python
# STRIPPING LEADING & TRAILING SPACES FROM ALL TEXT COLUMNS
text_cols = df_clean.select_dtypes(include=["object", "string"]).columns

for col in text_cols: df_clean[col] = df_clean[col].str.strip()
```

```python
# CHECKING OF MISSING VALUES
print(df_clean.isnull().sum())
```

```python
# FILLING THE MISSING VALUES IN 'AI ADOPTION LEVEL' COLUMN TO 'NONE'

df_clean['AI_Adoption_Level'] = df_clean['AI_Adoption_Level']. fillna('None')
```

```python
"""
1. Which tech domain has the highest average valuation?
"""

# Group by domain, calculate average valuation, and sort highest to lowest
domain_valuation  = (df_clean.groupby('Domain') 
                     ['Valuation_USD_Millions']
                     .mean()
                     .reset_index()
                     .sort_values(by='Valuation_USD_Millions',
                      ascending=False)
)

print(domain_valuation)
```

```python
"""
2. Which funding stage secures the highest average total funding?
"""

# Group by funding stage, calculate average total funding , and sort highest to lowest
funding_stage_summary = (df_clean.groupby('Funding_Stage')
                         ['Total_Funding_USD_Millions']
                         .mean()
                         .reset_index()
                         .sort_values(by='Total_Funding_USD_Millions',
                          ascending=False)       
)

print(domain_valuation)
```

```python
"""
3. Which AI adoption level has the highest total revenue ARR?
"""

# Group by AI adoption level, calculate total revenue, and sort highest to lowest
ai_revenue_summary = (df_clean.groupby("AI_Adoption_Level")
                     ['Revenue_ARR_Millions']
                     .sum()
                     .reset_index()
                     .sort_values(by='Revenue_ARR_Millions',
                      ascending=False)
)
print(ai_revenue_summary)
```

```python
"""
4. Which country has accumulated the most total funding?
"""
# Group by Country and calculate the total funding sum
country_total_funding = (df_clean.groupby("Country")
                     ['Total_Funding_USD_Millions']
                     .sum()
                     .reset_index()
                     .sort_values(by='Total_Funding_USD_Millions',
                      ascending=False)
)
print(country_total_funding)
```

```python
"""
5. Which tech domains have the highest number of startups running out of cash in under 6 months?
"""

# 5a. Define a  function with an if/else condition 
def flag_cash_risk(months):
    if months < 6:
        return 'Danger Zone'
    else:
        return 'Safe'

# 5b. Transform the dataframe by creating a new column 
df_clean['Risk_Status'] = df_clean['Runway_Months_2024'].apply(flag_cash_risk)

# 5c. Filter for the danger zone and count them by domain 
danger_summary = (df_clean[df_clean['Risk_Status']== 'Danger Zone']
                  .groupby('Domain')
                 ['Risk_Status']
                  .count()
                  .reset_index(name='Danger_Count')
                  .sort_values(by='Danger_Count',
                   ascending=False)
)
print(danger_summary)
```

```python
"""
6. Which startups are actually getting a return on their cash?
"""
# Create a new calculated column called Capital_Efficiency by dividing Revenue ARR by Total Funding
df_clean['Capital_Efficiency'] =df_clean['Revenue_ARR_Millions'] / df_clean['Total_Funding_USD_Millions']
```

```python
"""
7. Are well-funded startups safer or do they burn cash fast?
"""
# 7a. Define the labels for your funding tiers
funding_labels = ['Low Funding', 'Medium Funding', 'High Funding']

#7b. Create the new categorical column using automatic percentile splitting (qcut)
df_clean['Funding_Scale'] = pd.qcut(
                                df_clean['Total_Funding_USD_Millions'],
                                q=3,
                                labels=funding_labels
)


#7c. Group by funding scale to see where the Danger Zone is most concentrated
founding_risk_summary = (df_clean[df_clean['Risk_Status'] == 'Danger Zone']
                         .groupby('Funding_Scale', observed=False)['Risk_Status']
                         .count()
                         .reset_index(name='Danger_Count')
                         .sort_values(by='Danger_Count', ascending=False)
)

print("--- Danger Zone Breakdown  by Funding Scale---")
print(founding_risk_summary)
```

```python
"""
8. Does adopting AI actually extend a startup's runaway?
"""
# 8a. Group by AI adoption and Risk Status to evaluate survival rates
ai_survival_summary = (df_clean.groupby(['AI_Adoption_Level', 'Risk_Status'],
                                        observed=False)
                                        .size()
                                        .reset_index (name='Company_Count')
)

# 8b. Pivot the table so safe and Danger Zone  sit side-by-side 
ai_pivot = ai_survival_summary.pivot(index='AI_Adoption_Level',
                               columns='Risk_Status',
                               values='Company_Count'
).reset_index()

print("--- AI Adoption Level vs. Startup---")
print(ai_pivot)
```

```python
"""
KPIs
"""

# 1. Total Companies
total_companies = len(df_clean)
print(f"Total Companies Analyzed: {total_companies:,}")


# 2. Total Market Funding
total_market_funding = df_clean['Total_Funding_USD_Millions'].sum()
print(f"Total Market Funding: {total_market_funding:,.2f}")


# 3. Market Burn Risk Ratio (% in Danger Zone)
danger_count = len(df_clean[df_clean['Risk_Status'] == 'Danger Zone'])
burn_risk_ratio = (danger_count / total_companies) * 100
print(f"Market Burn Risk Ratio: {burn_risk_ratio:.2f}% of startups are in the Danger Zone.")

# 4. Average Funding by Risk Status
avg_funding_by_risk = (df_clean.groupby('Risk_Status', observed=False)
                       ['Total_Funding_USD_Millions']
                       .mean()
                       .reset_index(name='Average Funding')
                       )


print("---Average Funding by Risk Status---")
print(avg_funding_by_risk)

# 5. AI Adoption Survival Rate (% of Safe Companies within each AI level)
ai_totals = df_clean.groupby('AI_Adoption_Level', observed=False).size()

ai_safe_counts = df_clean[df_clean['Risk_Status'] == 'Safe'].groupby('AI_Adoption_Level',
                                    observed=False).size()

ai_survival_rate  = (ai_safe_counts / ai_totals * 100).reset_index(name='Safe_Percentage')

print("---AI Adoption Survival Rate (%)---")
print(ai_survival_rate)
```

```python
"""
VISUALIZATION
"""

import matplotlib.pyplot as plt

# Create an executive summary panel featuring clean dashboard metric cards
fig = plt.figure(figsize=(14, 8))
gs = fig.add_gridspec(3, 2, height_ratios=[1, 1, 1])

# --- Card 1: Total Companies ---
ax1 = fig.add_subplot(gs[0, 0])
ax1.axis('off')
ax1.text(0.5, 0.5, f"Total Companies Analyzed\n{total_companies:,}", 
             fontsize=12, fontweight='bold', ha='center', va='center',
             bbox=dict(boxstyle='round,pad=0.8', facecolor='#f8f9fa', edgecolor='#bdc3c7', linewidth=1.5))

# --- Card 2: Total Market Funding ---
ax2 = fig.add_subplot(gs[0, 1])
ax2.axis('off')
ax2.text(0.5, 0.5, f"Total Market Funding\n${total_market_funding:,.2f}M", 
             fontsize=12, fontweight='bold', ha='center', va='center',
             bbox=dict(boxstyle='round,pad=0.8', facecolor='#f8f9fa', edgecolor='#bdc3c7', linewidth=1.5))

# --- Card 3: Market Burn Risk Ratio ---
ax3 = fig.add_subplot(gs[1, 0])
ax3.axis('off')
ax3.text(0.5, 0.5, f"Market Burn Risk Ratio\n{burn_risk_ratio:.2f}% (Danger Zone)", 
             fontsize=11, fontweight='bold', ha='center', va='center',
             bbox=dict(boxstyle='round,pad=0.7', facecolor='#fadbd8', edgecolor='#e74c3c', linewidth=1.5))

# --- Card 4: Average Funding Summary ---
safe_avg = avg_funding_by_risk[avg_funding_by_risk.iloc[:, 0] == 'Safe'].iloc[0, 1]
danger_avg = avg_funding_by_risk[avg_funding_by_risk.iloc[:, 0] == 'Danger Zone'].iloc[0, 1]
ax4 = fig.add_subplot(gs[1, 1])
ax4.axis('off')
ax4.text(0.5, 0.5, f"Average Funding by Risk Status\nSafe: ${safe_avg:,.1f}M  |  Danger: ${danger_avg:,.1f}M", 
             fontsize=10, fontweight='bold', ha='center', va='center',
             bbox=dict(boxstyle='round,pad=0.7', facecolor='#e8f8f5', edgecolor='#2ecc71', linewidth=1.5))

# --- Card 5: AI Adoption Survival Rate Summary ---
overall_survival = ai_survival_rate['Safe_Percentage'].mean()
ax5 = fig.add_subplot(gs[2, :])
ax5.axis('off')
ax5.text(0.5, 0.5, f"Average AI Adoption Survival Rate (Across All Levels)\n{overall_survival:.2f}% Overall Safe Survival", 
             fontsize=11, fontweight='bold', ha='center', va='center',
             bbox=dict(boxstyle='round,pad=0.7', facecolor='#ebf5fb', edgecolor='#3498db', linewidth=1.5))

plt.suptitle('Tech Startup Performance Dashboard', fontsize=14, fontweight='bold', y=0.95)
plt.tight_layout()
plt.show()
```
